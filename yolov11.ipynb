```python
%matplotlib inline

import os
import random
import matplotlib.pyplot as plt
from PIL import Image

# Function to select and plot random images
def select_and_plot_random_images(source_dir, num_samples, grid_size=(3, 3), figsize=(12,10)):
    """
    Select a given number of random images from subdirectories of the source directory
    and plot them in a grid using matplotlib.

    Parameters:
        source_dir (str): Path to the source directory containing subdirectories of images.
        num_samples (int): Number of random samples to select.
        grid_size (tuple): Grid size for plotting (rows, columns).
    """
    all_images = []

    # Collect all image file paths
    for root, _, files in os.walk(source_dir):
        for file in files:
            if file.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".gif")):
                all_images.append(os.path.join(root, file))

    if num_samples > len(all_images):
        raise ValueError("Number of samples exceeds the total number of images available.")

    # Randomly select images
    selected_images = random.sample(all_images, num_samples)

    # Plot the images in a grid
    fig, axes = plt.subplots(grid_size[0], grid_size[1], figsize=figsize)
    axes = axes.flatten()

    for ax, img_path in zip(axes, selected_images):
        img = Image.open(img_path)
        ax.imshow(img)
        ax.axis('off')

    # Hide any extra axes
    for ax in axes[len(selected_images):]:
        ax.axis('off')

    plt.tight_layout()
    plt.show()

source_directory = r"C:\Users\101081\Desktop\Advc.Topics\Project\Foggy_CityScapes_Dataset\data"
number_of_samples = 9 
grid_dimensions = (3, 3)

select_and_plot_random_images(source_directory, number_of_samples, grid_dimensions, figsize=(15,8))
```

```python
import cv2
import numpy as np

def dark_channel_prior(image, window_size=15):
    # Minimum across color channels
    min_channel = np.min(image, axis=2)
    # Apply erosion to get the dark channel
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (window_size, window_size))
    dark_channel = cv2.erode(min_channel, kernel)
    return dark_channel

def estimate_atmospheric_light(image, dark_channel, top_percent=0.001):
    # Flatten the dark channel
    h, w = dark_channel.shape
    num_pixels = h * w
    num_brightest = int(max(num_pixels * top_percent, 1))
    
    # Get indices of top dark_channel values
    dark_vec = dark_channel.ravel()
    image_vec = image.reshape(num_pixels, 3)
    indices = np.argsort(dark_vec)[-num_brightest:]
    
    # Atmos. light is mean of these brightest pixels in original image
    atmo_light = np.mean(image_vec[indices], axis=0)
    return atmo_light

def estimate_transmission(image, atmospheric_light, omega=0.95, window_size=15):
    # Normalize by atmospheric light
    norm_image = image / atmospheric_light
    # Compute dark channel of normalized image
    dark_channel_norm = dark_channel_prior(norm_image, window_size)
    # Estimate transmission
    transmission = 1 - omega * dark_channel_norm
    return transmission

def guided_filter(I, p, radius, eps):
    # Guided filter to refine the transmission map
    mean_I = cv2.boxFilter(I, cv2.CV_64F, (radius, radius))
    mean_p = cv2.boxFilter(p, cv2.CV_64F, (radius, radius))
    mean_Ip = cv2.boxFilter(I * p, cv2.CV_64F, (radius, radius))
    cov_Ip = mean_Ip - mean_I * mean_p

    mean_II = cv2.boxFilter(I * I, cv2.CV_64F, (radius, radius))
    var_I = mean_II - mean_I * mean_I

    a = cov_Ip / (var_I + eps)
    b = mean_p - a * mean_I

    mean_a = cv2.boxFilter(a, cv2.CV_64F, (radius, radius))
    mean_b = cv2.boxFilter(b, cv2.CV_64F, (radius, radius))

    q = mean_a * I + mean_b
    return q

def recover_image(image, transmission, atmospheric_light, t_min=0.1):
    # Ensure transmission isn’t too low
    transmission = np.maximum(transmission, t_min)
    # Expand dims for broadcasting
    transmission_3d = np.expand_dims(transmission, axis=2)
    J = (image - atmospheric_light) / transmission_3d + atmospheric_light
    return np.clip(J, 0, 1)

def dehaze(image_path, from_path = True):

    image = image_path
    # Read image
    if (from_path):
        image = cv2.imread(image_path)
    # Convert BGR to RGB and normalize
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) / 255.0

    # Steps:
    dark = dark_channel_prior(image)
    A = estimate_atmospheric_light(image, dark)
    raw_t = estimate_transmission(image, A)

    # Refine transmission using guided filter
    gray = cv2.cvtColor((image * 255).astype(np.uint8), cv2.COLOR_RGB2GRAY) / 255.0
    refined_t = guided_filter(gray, raw_t, radius=60, eps=1e-3)

    # Recover haze-free image
    dehazed = recover_image(image, refined_t, A)
    dehazed_bgr = cv2.cvtColor((dehazed * 255).astype(np.uint8), cv2.COLOR_RGB2BGR)
    return dehazed_bgr

def gamma_correction(image, gamma):
    # Avoid division by zero and ensure gamma is positive
    gamma = max(gamma, 1e-8)
    
    # Compute the inverse gamma value
    inv_gamma = 1.0 / gamma

    # Build a lookup table to speed up the pixel value transformation
    # The table maps each input pixel value [0..255] to its gamma-corrected output
    table = np.array([(i / 255.0) ** inv_gamma * 255 for i in range(256)]).astype("uint8")

    # Apply the lookup table to the input image
    corrected = cv2.LUT(image, table)
    return corrected

def apply_clahe(img, clipLimit=2.0, tileGridSize=(8,8)):
    # Convert from BGR to Lab color space
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    
    # Split into L, a, b channels
    l, a, b = cv2.split(lab)
    
    # Create CLAHE object
    clahe = cv2.createCLAHE(clipLimit=clipLimit, tileGridSize=tileGridSize)
    
    # Apply CLAHE on the L channel
    l_clahe = clahe.apply(l)
    
    # Merge the channels back
    lab_clahe = cv2.merge((l_clahe, a, b))
    
    # Convert back to BGR color space
    img_clahe = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2BGR)
    return img_clahe
```

```python
def preprocess_images(target_dir, output_dir, apply_dehaze=True, gcorr=True, clahe=True, edenh=True, gamma=2.2):
    """
    Applies gamma correction to all .png images in target_dir and saves them to output_dir,
    preserving the directory structure.
    """
    for root, dirs, files in os.walk(target_dir):
        # Construct relative path from target directory to current directory
        rel_path = os.path.relpath(root, target_dir)
        # Create corresponding directory in output directory
        output_sub_dir = os.path.join(output_dir, rel_path)
        if not os.path.exists(output_sub_dir):
            os.makedirs(output_sub_dir, exist_ok=True)

        for file_name in files:
            if file_name.lower().endswith('.png'):
                target_file_path = os.path.join(root, file_name)
                output_file_path = os.path.join(output_sub_dir, file_name)

                # Open the image
                with Image.open(target_file_path) as img:
                    # Convert image to RGB if not already
                    img = img.convert('RGB')
                    # Apply gamma correction
                    if (True == apply_dehaze):
                        img = dehaze(target_file_path)

                    if (True == gcorr):
                        img = gamma_correction(img, gamma)

                    if (True == clahe):
                        img = apply_clahe(img)

                    if (True == edenh):
                        pil_img = Image.fromarray(img)
                        img = np.array(pil_img.filter(ImageFilter.EDGE_ENHANCE_MORE))

                    # Save the corrected image
                    cv2.imwrite(output_file_path, img)

original_dataset_path = "C:/Users/101081/Desktop/Advc.Topics/Project/Foggy_CityScapes_Dataset/yolo_format"
processed_dataset_path = "C:/Users/101081/Desktop/Advc.Topics/Project/Foggy_CityScapes_Dataset/dehaze_gc_2_8"

preprocess_images(original_dataset_path, processed_dataset_path, apply_dehaze=True, gcorr=True, clahe=False, edenh=False, gamma=2.8)
```

```python
video_path = "./test.mp4"  # Replace with your video file
output_path = "./output_dehazed_model.mp4"

model = YOLO('./runs/detect/1_train_original/weights/best.pt')

# OpenCV VideoCapture
cap = cv2.VideoCapture(video_path)

# Retrieve properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Define the codec and create VideoWriter to save output
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # or 'mp4v' for .mp4
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

while True:
    ret, frame = cap.read()
    if not ret:
        break  # End of video

    # inf_frame = dehaze(frame, False)
    # frame = gamma_correction(frame, gamma=2.2)

    # Run inference
    results = model(frame)
    predictions = results[0].boxes if hasattr(results[0], 'boxes') else []
    
    # Draw bounding boxes and labels on the frame
    # Adjust the code below depending on how your YOLO model returns predictions
    for box in predictions:
        # Each 'box' typically has xyxy coordinates, confidence, and class
        # For Ultralytics YOLO:
        # box.xyxy -> (x1, y1, x2, y2)
        # box.conf -> confidence
        # box.cls  -> class ID
        
        x1, y1, x2, y2 = box.xyxy[0]
        conf = float(box.conf[0])
        cls_id = int(box.cls[0])

        if (conf  <= 0.6): continue
        
        # For class labels, if model.names is available (Ultralytics YOLOv11)
        class_name = model.names[cls_id] if model.names and cls_id in model.names else str(cls_id)
        
        # Convert to int for bounding box drawing
        x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
        
        # Draw rectangle
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        
        # Prepare label text
        label = f"{class_name}: {conf:.2f}"
        
        # Draw label background for clarity
        (text_width, text_height), baseline = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.5, 2)
        cv2.rectangle(
            frame, 
            (x1, y1 - text_height - 5), 
            (x1 + text_width, y1), 
            (0, 255, 0), 
            cv2.FILLED
        )
        
        # Put text above the rectangle
        cv2.putText(
            frame, 
            label, 
            (x1, y1 - 5), 
            cv2.FONT_HERSHEY_SIMPLEX, 
            0.5, 
            (0, 0, 0), 
            2
        )
    
    # Write the processed frame to output
    out.write(frame)

# Release everything
cap.release()
out.release()
cv2.destroyAllWindows()
```
